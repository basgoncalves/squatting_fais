```python
import numpy as np
import pandas as pd
import sys
print (sys.version)

import os
import msk_modelling_python as msk
```

```python
def header_mot(df,name):

        num_rows = len(df)
        num_cols = len(df.columns) 
        inital_time = df['time'].iloc[0]
        final_time = df['time'].iloc[-1]
        df_range = f'{inital_time} {final_time}'

        # return f'name {name}\nnRows={num_rows}\nnColumns={num_cols}\n \nendheader'
        return f'name {name}\ndatacolumns {num_cols}\ndatarows {num_rows}\nrange {df_range}\nendheader'

def csv_to_mot(emg_csv, columns = 'all'):
    '''
    '''
    
    emg_data = pd.read_csv(emg_csv)
    
    try:
        time = emg_data['time']
    except:
        time = emg_data['Time']

    # start time from new time point
    start_time = time.iloc[0]
    end_time = time.iloc[-1]

    num_samples = len(emg_data)
    new_time = np.linspace(start_time, end_time, num_samples)
    # why????? 

    # create new df for modification
    new_emg_data = pd.DataFrame()
    try:
        time = emg_data['time']
    except:
        time = emg_data['Time']

    new_emg_data['time'] = new_time
    new_emg_data['ground_force_vx'] = emg_data['Force_Fx1'] # creates new column in the new dataframe new_emg_data under 'ground_force_vx' with data from emg_data under 'Force_Fx1'
    new_emg_data['ground_force_vy'] = emg_data['Force_Fy1']
    new_emg_data['ground_force_vz'] = emg_data['Force_Fz1']

    new_emg_data['ground_force_px'] = emg_data['Force_Fx2']
    new_emg_data['ground_force_py'] = emg_data['Force_Fy2']
    new_emg_data['ground_force_pz'] = emg_data['Force_Fz2']

    new_emg_data['l_ground_force_vx'] = emg_data['Force_Fx3']
    new_emg_data['l_ground_force_vy'] = emg_data['Force_Fy3']
    new_emg_data['l_ground_force_vz'] = emg_data['Force_Fz3']

    new_emg_data['l_ground_force_px'] = emg_data['Force_Fx4']
    new_emg_data['l_ground_force_py'] = emg_data['Force_Fy4']
    new_emg_data['l_ground_force_pz'] = emg_data['Force_Fz4']

    new_emg_data['ground_torque_x'] = emg_data['Moment_Mx1']
    new_emg_data['ground_torque_y'] = emg_data['Moment_My1']
    new_emg_data['ground_torque_z'] = emg_data['Moment_Mz1']

    # new_emg_data['l_ground_torque_x'] = emg_data['Moment_Mx2']
    # new_emg_data['l_ground_torque_y'] = emg_data['Moment_My2']
    # new_emg_data['l_ground_torque_z'] = emg_data['Moment_Mz2']

    new_emg_data['l_ground_torque_x'] = emg_data['Moment_Mx3']
    new_emg_data['l_ground_torque_y'] = emg_data['Moment_My3']
    new_emg_data['l_ground_torque_z'] = emg_data['Moment_Mz3']

    # Define a new file path 
    new_file_path = os.path.join(emg_csv.replace('.csv', '.mot'))

    # Save the modified DataFrame
    new_emg_data.to_csv(new_file_path, index=False)  # index=False prevents adding an extra index column

    filename = os.path.basename(emg_csv)
    name = filename.split('.')[0]   # name like the actual filename
    # name = "processed_emg_signals" # previous version
    # save to mot
    header = header_mot(new_emg_data, name)

    mot_path = new_file_path.replace('.csv','.mot')
    with open(mot_path, 'w') as f:
        f.write(header + '\n')  
        # print column names 
        f.write('\t'.join(map(str, new_emg_data.columns)) + '\n')
        for index, row in new_emg_data.iterrows():
            f.write('\t'.join(map(str, row.values)) + '\n')  
    
    print(f"File saved: {mot_path}")
    
    return mot_path
```
